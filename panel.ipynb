# Análisis de ventas

```python
import json
with open("data.json", "r") as file:
    data = json.load(file)
```

## Ejercicio 1: Producto con más ingresos

```python
ingresos_producto = {}
for venta in data["ventas"]:
    producto = venta["producto"]
    importe = venta["cantidad"] * venta["precio_unitario"]
    
    if producto not in ingresos_producto:
        ingresos_producto[producto] = 0
        
    ingresos_producto[producto] = ingresos_producto[producto] + importe
    
producto_mas_ingresos = ""
max_ingresos = 0

for producto in ingresos_producto:
    if ingresos_producto[producto] > max_ingresos:
        max_ingresos = ingresos_producto[producto]
        producto_mas_ingresos = producto
        
print("Producto con más ingresos:", producto_mas_ingresos)
print("Ingresos:", max_ingresos)
```

## Ejercicio 2

```python
vendedor = "Ana"
ventas_filtradas = []


for venta in data ["ventas"]:
    if venta["vendedor"] == vendedor:
        ventas_filtradas.append(venta)
        

print(ventas_filtradas)
```

## Ejercicio 3

```python
total_facturado = 0
numero_ventas = 0

for venta in data["ventas"]:
    importe = venta["cantidad"] * venta["precio_unitario"]
    total_facturado = total_facturado + importe
    numero_ventas = numero_ventas + 1
    
ticket_medio = total_facturado / numero_ventas


print("Ticket medio:", ticket_medio)
```
